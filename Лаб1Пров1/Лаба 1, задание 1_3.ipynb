## Задача 1
Написать программу, которая будет выводить на экран числа Фибоначчи, которые меньше переданного числа.

```python
n = int(input("Введите число: "))
previous, current = 1, 1
condition = True

print(f'Числа Фибоначчи, которые меньше, чем переданное число ({n}): ', end='')
while condition:
    if current + previous < n:
        previous, current = current, previous + current
        print(current, end=' ')
    else:
        condition = False
```

```python
def func1(z):
    res = []
    for item in z:
        if not res.count(item):
            res.append(item)
    return res

k = func1([1,3,4,2,1,3,4,2,5])

for item in k:
        print(item)
```

## Задача 2
Напишите программу, которая на вход принимает список чисел и возвращает только уникальные значения из этого списка.

```python
#я не понял задание. сделал оба варианта

lst_ = input("Введите числа через пробел: ")

#Первый - возвращает элементы без повторения
print('Отладочный вывод введённого списка:', lst_)
lst_ = lst_.split()
numbers = sorted(list(set(lst_)))
print("\nВывод всех элементов списка без повторения: ", numbers)

#Второй - вовращает только уникальные элементы из списка, повторки не выводятся
seen_check = {}
answer = []
for num in lst_:
    if num in seen_check:
        seen_check[num] += 1
    else:
        seen_check[num] = 1

for num in seen_check:
    if seen_check[num] == 1:
        answer.append(num)

print("\nВывод только уникальных элементов из списка: ", answer)
```

## Задача 3
Напишите программу, которая переводит температуру в градусах Фаренгейта в градусы Цельсия.


```python
fahrenheit = float(input("Введите температуру в градусах Фаренгейта: "))
celsium = round((fahrenheit - 32) * 5 / 9, 2) #округлил встроенной функией round для адекватного глазу вывода
print(f'{fahrenheit} °F = {celsium} °C')
```

## Задача 4
Напишите программу, которая проверяет, является ли строка палиндромом без использования стандартных методов.

```python
sentence = input("Введите строку для проверки её на палиндром: ")
sentence = "".join(sentence.lower().split())
condition = True

#print (sentence)   #Вывод отформатированной для последующей проверки строки

if len(sentence) > 1:
    for i in range (len(sentence) // 2):
        if sentence[i] != sentence[len(sentence) - i -1]:
            condition = False
    if condition == True:
        print("Эта строка - палиндром")
    else:
        print("Эта строка не палиндром")
else:
    print("Строка должна содержать более одного символа")

```
