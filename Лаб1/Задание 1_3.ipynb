## Задача 1
Написать программу, которая будет выводить на экран числа Фибоначчи, которые меньше переданного числа.

```python
stop = int(input("Введите край для последовательности Фибоначчи"))
a, b = 1, 1
while (a < stop):
    print(a)
    a, b = b, a + b

```

## Задача 2
Напишите программу, которая на вход принимает список чисел и возвращает только уникальные значения из этого списка.

```python
numbers = [int(x) for x in input("Введите числа через пробел: ").split()]

for i in numbers:
    if (numbers.count(i) == 1):
        print(i)
```

## Задача 3
Напишите программу, которая переводит температуру в градусах Фаренгейта в градусы Цельсия.


```python
F = float(input("Введите температуру в градусах Фаренгейта: "))
C = (F - 32) * 5 / 9
print(f"Температура в градусах Цельсия: {C:.2f}")
```

## Задача 4
Напишите программу, которая проверяет, является ли строка палиндромом без использования стандартных методов.

```python
text = input("Введите строку для проверки: ")

text = text.replace(" ", "").lower() 

is_palindrome = True
start = 0
end = len(text) - 1

while start < end:
    if text[start] != text[end]:
        is_palindrome = False
        break
    start += 1
    end -= 1

if is_palindrome:
    print("Строка является палиндромом")
else:
    print("Строка не является палиндромом")
```
